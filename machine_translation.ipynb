```python
examples, metadata= tfds.load("ted_hrlr_translate/pt_to_en", as_supervised= True, with_info= True) 
train_examples, validation_example= examples["train"], examples["validation"]

for portugese, english in train_examples.batch(3).take(1): 
    print("Portugese example: ")
    for pt in portugese.numpy(): 
        print(pt.decode("utf-8"))
    
    print("English examples: ")    
    for eng in english.numpy(): 
        print(eng.decode("utf-8"))
        
```

### Setting up Tokenizer 

```python
model_name= "ted_hrlr_translate_pt_en_converter"
tf.keras.utils.get_file( f"{model_name}.zip", f'https://storage.googleapis.com/download.tensorflow.org/models/{model_name}.zip', 
                        cache_dir=".", cache_subdir='', extract= True)
```

```python
tokenizers= tf.saved_model.load(model_name)
```

```python
[items for items in dir(tokenizers.en) if not items.startswith("_")]
```

```python
print("This is a batch of strings")
for en in english.numpy(): 
    print(en.decode("utf-8"))
```

```python
encoded= tokenizers.en.tokenize(english)

print("This is a padded batch of token ID: ")
for row in encoded.to_list(): 
    print(row)
```

```python
round_trip= tokenizers.en.detokenize(encoded)

print("This is a human readable form")
for line in round_trip.numpy(): 
    print(line.decode("utf-8"))
```
